## 1. JSON: a structured data format
JSON (JavaScript Object Notation) represents data as text. It is widely used in API responses because programs can parse its objects, arrays, and scalar values.

A JSON object resembles a Python dictionary but is not Python code. JSON uses double quotes, lowercase boolean values, and `null`. Comments and trailing commas are not part of standard JSON.

| JSON type | Python type after parsing | Example |
|---|---|---|
| object | `dict` | `{"city": "Houghton"}` |
| array | `list` | `[10, 12, 14]` |
| string | `str` | `"temperature"` |
| number | `int` or `float` | `12.5` |
| boolean | `bool` | `true` → `True` |
| null | `NoneType` | `null` → `None` |

### Example 1 — Parse a JSON string
`json.loads()` parses a string. Use keys for dictionary fields and integer indices for list elements.

```python
book_text = """{
    "title": "Data Science Book", "author": "Joel Grus",
    "publicationYear": 2014,
    "topics": ["Python", "statistics", "data science"],
    "available": true, "rating": null
}"""
book = json.loads(book_text)
print(type(book_text).__name__, "→", type(book).__name__)
print(book["title"])
print(book["topics"][1])
print("data science" in book["topics"])
print(book["available"], book["rating"])
```

### Example 2 — Serialize a Python object
`json.dumps()` converts a Python object into JSON text. Serialization is useful when sending data to a service or saving results. `indent=2` makes the text easier to read.

The extra **s** in `loads` and `dumps` means string. Do not use `eval()` to parse unfamiliar JSON.

```python
book["topics"].append("APIs")
serialized_book = json.dumps(book, indent=2, allow_nan=False)
print(serialized_book)
restored_book = json.loads(serialized_book)
assert restored_book == book
```

### Example 3 — Nested records
A common structure is a dictionary containing a list of records. Here `persons` is a list, and each person is a dictionary containing another dictionary called `location`.

Read the structure from the outside inward: **object → persons list → person → location → city**.

```python
people_text = """{
  "persons": [
    {"id": "001", "name": "Brian", "location": {"city": "Seattle", "country": "USA"}},
    {"id": "002", "name": "David", "location": {"city": "Amsterdam", "country": "Netherlands"}}
  ]
}"""
people = json.loads(people_text)
print(people["persons"][0]["location"]["city"])
for person in people["persons"]:
    print(person["name"], person["location"]["city"])
people_df = pd.json_normalize(people["persons"])
display(people_df)
```

`pd.json_normalize()` flattens nested dictionaries into columns such as `location.city`. Keep identifiers like `"001"` as strings if leading zeros matter.

**Common mistake:** `people['persons']['name']` fails because `persons` is a list. Select an item first or iterate over all items.

### Example 4 — JSON files
`json.dump()` writes to an open file; `json.load()` reads from an open file.

```python
json_path = output_dir / "people.json"
with json_path.open("w", encoding="utf-8") as f:
    json.dump(people, f, indent=2)
with json_path.open(encoding="utf-8") as f:
    people_from_file = json.load(f)
assert people_from_file == people
print("Loaded people:", len(people_from_file["persons"]))
```

### Missing and invalid data
A present field with `null` differs from a missing key. Use direct indexing for required fields and `.get()` for genuinely optional fields. A missing temperature should not become zero, because zero is a valid measurement.

Standard JSON has no `NaN` or infinity. Use `allow_nan=False` when serializing if you need to reject those nonstandard numeric values.

```python
record = {"temperature_c": None}
print("Field present:", "temperature_c" in record)
print("Value:", record["temperature_c"])
print("Optional humidity:", record.get("humidity", "not supplied"))
try:
    json.loads('{"available": True}')
except json.JSONDecodeError as error:
    print("Invalid JSON:", error.msg)
```

## 2. Small transformations with `lambda` and `map()`
### Lambda functions
A lambda is an anonymous function containing one expression. It is useful for a short transformation or a sorting key. Use a named `def` function when the logic needs multiple steps, documentation, or reuse.

Syntax: `lambda arguments: expression`

```python
def add_ten(x):
    return x + 10

add_ten_lambda = lambda x: x + 10
print(add_ten(5))
print(add_ten_lambda(5))

multiply = lambda x, y: x * y
print(multiply(3, 4))
```

### Sorting records with a lambda
The `key` function extracts the value used for sorting. This pattern will help us rank API records without changing their other fields.

```python
measurements = [
    {"city": "Houghton", "temperature_c": 10.0},
    {"city": "Marquette", "temperature_c": 12.0},
    {"city": "Duluth", "temperature_c": 8.0}
]
ordered = sorted(measurements, key=lambda row: row["temperature_c"], reverse=True)
display(pd.DataFrame(ordered))
```

### `map()` applies a function to each item
In Python 3, `map()` returns an iterator. Convert it to a list to inspect all results. Once consumed, that iterator has no remaining values.

Here we convert Celsius to Fahrenheit using **F = 1.8C + 32**.

```python
temperatures_c = [0, 10, 20]
converted = map(lambda value: 1.8 * value + 32, temperatures_c)
print(type(converted).__name__)
print(list(converted))
print("After consumption:", list(converted))

# A list comprehension expresses the same transformation.
print([1.8 * value + 32 for value in temperatures_c])
```

### Extracting fields and transforming pandas columns
`map()` can extract fields from a list of dictionaries. `Series.map()` is a pandas method: it maps each Series value using a function or mapping. It is distinct from Python's built-in `map()`.

For arithmetic on an entire numeric Series, direct vectorized expressions are usually clearer.

```python
cities = list(map(lambda row: row["city"], measurements))
print(cities)
measurement_df = pd.DataFrame(measurements)
measurement_df["temperature_f"] = measurement_df["temperature_c"].map(lambda x: 1.8 * x + 32)
measurement_df["temperature_f_vectorized"] = 1.8 * measurement_df["temperature_c"] + 32
display(measurement_df)
```

### Transformation practice
- Use `map()` to extract the names from the JSON people records.
- Sort the measurement records alphabetically by city.
- Rewrite the Celsius conversion as a named function.

**Discussion:** Use a lambda for a simple expression; use a named function for complex logic. Neither lambda nor map automatically makes code faster. Prioritize clarity and choose vectorized pandas operations for ordinary column arithmetic.

## 3. Web API fundamentals
An API specifies how programs interact. A web API commonly accepts HTTP requests and returns structured data. When a documented API supplies the needed data, it is often more convenient than parsing a webpage's HTML.

An API is not a guarantee of clean data, complete coverage, or unrestricted use. Read variable definitions and access terms.

| Component | Meaning |
|---|---|
| Endpoint | URL identifying a resource or operation |
| Method | `GET` retrieves; `POST` submits data for an operation |
| Parameters | Filters, dates, locations, or page controls |
| Headers | Representation preferences and sometimes credentials |
| Status | Indicates whether the HTTP request succeeded |
| Body | Returned data, often JSON |

**Basic pattern:** `requests.get(url, params=params, timeout=30)` → `raise_for_status()` → `.json()`.

Use `params=` to encode query values. Check status before parsing: error responses can also be valid JSON. A timeout limits connection/read inactivity; it is not a strict overall workflow deadline.

### Three different response structures
| API | Response structure | DataFrame row |
|---|---|---|
| NASA NeoWs | Dictionary of dates, each containing asteroid records | One asteroid close-approach event |
| Open-Meteo | Dictionary containing aligned daily arrays | One local date at one location |
| World Bank | List containing page metadata and observations | One country–year–indicator |

The conversion depends on the response structure. Inspect the JSON and define the row meaning before constructing a table.

## 4. NASA API: near-Earth asteroid approaches
[Official NASA API documentation](https://api.nasa.gov/) — select **Asteroids — NeoWs**.

**Question:** Which asteroid approaches are reported for a fixed date range, and how do their estimated size and miss distance compare?

NeoWs provides near-Earth object records based on approach dates. Use a short window within the feed endpoint's allowed maximum range. The public `DEMO_KEY` is suitable for small demonstrations but has a shared, limited quota. Do not repeatedly rerun the request cell during class.

### Step 1 — Request the data
The endpoint is `/neo/rest/v1/feed`. Parameters select the start and end dates and supply a key. A personal key, if configured, remains in the environment; do not print the full request URL because NASA places the key in its query string.

```python
nasa_url = "https://api.nasa.gov/neo/rest/v1/feed"
nasa_dates = {"start_date": "2025-10-01", "end_date": "2025-10-03"}
nasa_key = os.getenv("NASA_API_KEY") or "DEMO_KEY"
nasa_response = requests.get(
    nasa_url, params={**nasa_dates, "api_key": nasa_key}, timeout=30
)
print("NASA status:", nasa_response.status_code)
if not nasa_response.ok:
    raise RuntimeError(f"NASA returned HTTP {nasa_response.status_code}; check the key, quota, and dates")
nasa = nasa_response.json()
nasa_retrieved_at = datetime.now(timezone.utc).isoformat()
```

### Step 2 — Inspect the nested JSON
`near_earth_objects` maps date strings to lists of objects. Each object contains an estimated-diameter dictionary and a list of close-approach records.

Print a selected subset rather than the entire response. This also avoids displaying response links that may contain a key.

```python
print("Top-level keys:", list(nasa.keys()))
print("Reported element count:", nasa["element_count"])
print("Date groups:", list(nasa["near_earth_objects"].keys()))
objects = [obj for daily_objects in nasa["near_earth_objects"].values() for obj in daily_objects]
if objects:
    first = objects[0]
    print("Example:", first["name"])
    print("Diameter estimate (km):", first["estimated_diameter"]["kilometers"])
    print("Approach records:", len(first["close_approach_data"]))
```

### Step 3 — Extract records into a DataFrame
Each row represents an asteroid approach to Earth within the requested dates. Iterate over `close_approach_data` rather than assume it has exactly one entry. Use asteroid ID plus approach timestamp as the event key.

Diameter estimates are ranges. The midpoint below is a convenient summary of that range, not a precise measurement. API velocity and distance fields are numeric strings, so explicitly convert them.

```python
asteroid_rows = []
for asteroid in objects:
    diameter = asteroid["estimated_diameter"]["kilometers"]
    for approach in asteroid["close_approach_data"]:
        date = approach["close_approach_date"]
        if approach["orbiting_body"] != "Earth" or not nasa_dates["start_date"] <= date <= nasa_dates["end_date"]:
            continue
        asteroid_rows.append({
            "asteroid_id": asteroid["id"], "name": asteroid["name"],
            "approach_date": date,
            "approach_time_ms": approach["epoch_date_close_approach"],
            "diameter_min_km": diameter["estimated_diameter_min"],
            "diameter_max_km": diameter["estimated_diameter_max"],
            "velocity_km_s": float(approach["relative_velocity"]["kilometers_per_second"]),
            "miss_distance_km": float(approach["miss_distance"]["kilometers"]),
            "potentially_hazardous": asteroid["is_potentially_hazardous_asteroid"]
        })
asteroid_df = pd.DataFrame(asteroid_rows, columns=[
    "asteroid_id", "name", "approach_date", "approach_time_ms",
    "diameter_min_km", "diameter_max_km", "velocity_km_s",
    "miss_distance_km", "potentially_hazardous"
])
asteroid_df["approach_date"] = pd.to_datetime(asteroid_df["approach_date"])
asteroid_df["diameter_mid_km"] = (asteroid_df["diameter_min_km"] + asteroid_df["diameter_max_km"]) / 2
if asteroid_df.duplicated(["asteroid_id", "approach_time_ms"]).any():
    raise ValueError("Duplicate asteroid approach events")
print("Approach rows:", len(asteroid_df))
display(asteroid_df.head(10))
```

### Step 4 — Rank and summarize
Use the same sorting-key pattern introduced with lambda. Rank event records by miss distance, then display the closest five. Group by date to count retrieved approach events.

```python
closest = sorted(asteroid_rows, key=lambda row: row["miss_distance_km"])[:5]
display(pd.DataFrame(closest)[["name", "approach_date", "miss_distance_km", "velocity_km_s"]]
        if closest else asteroid_df.head())
approach_counts = asteroid_df.groupby("approach_date").size().rename("approach_count")
display(approach_counts.to_frame())
if not approach_counts.empty:
    approach_counts.plot.bar(rot=30)
    plt.ylabel("Retrieved approach events")
    plt.title("NASA near-Earth asteroid approaches")
    plt.tight_layout()
    plt.show()
```

### Step 5 — Interpret and practice
- Filter the DataFrame to objects marked potentially hazardous.
- Compare sorting by size with sorting by miss distance. Does the largest object have the closest approach?
- Explain why the NASA element count and extracted approach-row count need not always be identical.

**Interpretation:** The feed groups objects while the table represents events; an object may have multiple approaches, and our filters define which events are retained. “Potentially hazardous” is a classification, not a prediction of an imminent collision. A short retrieved interval cannot describe all asteroids or their long-term risk.

## 5. Weather API: Houghton daily weather
[Open-Meteo historical weather documentation](https://open-meteo.com/en/docs/historical-weather-api)

**Question:** How did daily maximum temperature and precipitation vary near Houghton during October 2025?

This example requests a fixed historical month. The archive provides reanalysis/model weather data; it should not be described as measurements from a particular local weather station.

### Step 1 — Specify location, date range, and variables
Latitude and longitude identify a location. `daily` selects variables. `timezone` defines the daily calendar periods.

```python
weather_url = "https://archive-api.open-meteo.com/v1/archive"
weather_params = {
    "latitude": 47.12, "longitude": -88.57,
    "start_date": "2025-10-01", "end_date": "2025-10-31",
    "daily": "temperature_2m_max,precipitation_sum",
    "timezone": "America/Detroit"
}
r = requests.get(weather_url, params=weather_params, timeout=30)
r.raise_for_status()
weather = r.json()
print("Response keys:", list(weather.keys()))
print("Units:", weather["daily_units"])
print("Timezone:", weather["timezone"])
weather_retrieved_at = datetime.now(timezone.utc).isoformat()
```

### Step 2 — Understand the response structure
The `daily` object contains aligned arrays. The first date, first temperature, and first precipitation value describe the same day. This differs from GitHub's list of record dictionaries.

Inspect the first three entries before creating the table.

```python
daily = weather["daily"]
for key in ["time", "temperature_2m_max", "precipitation_sum"]:
    print(key, ":", daily[key][:3])
lengths = [len(daily[key]) for key in ["time", "temperature_2m_max", "precipitation_sum"]]
if len(set(lengths)) != 1:
    raise ValueError("Daily arrays have different lengths")
```

### Step 3 — Build a DataFrame
Each row represents one local calendar day at the selected location. Convert the date column and retain missing values.

```python
weather_df = pd.DataFrame(daily).rename(columns={
    "time": "date", "temperature_2m_max": "max_temperature",
    "precipitation_sum": "precipitation"
})
weather_df["date"] = pd.to_datetime(weather_df["date"])
display(weather_df.head())
print("Number of days:", len(weather_df))
print("Missing values:")
print(weather_df.isna().sum())
```

### Step 4 — Visualize and summarize
Use separate panels because temperature and precipitation have different units. Always read the units from the response rather than assume them.

```python
fig, axes = plt.subplots(2, 1, figsize=(9, 6), sharex=True)
axes[0].plot(weather_df["date"], weather_df["max_temperature"], marker="o")
axes[0].set_ylabel("Max temperature (" + weather["daily_units"]["temperature_2m_max"] + ")")
axes[1].bar(weather_df["date"], weather_df["precipitation"])
axes[1].set_ylabel("Precipitation (" + weather["daily_units"]["precipitation_sum"] + ")")
axes[1].set_xlabel("Local date")
fig.suptitle("Houghton-area weather — October 2025")
fig.autofmt_xdate()
plt.tight_layout()
plt.show()
print("Average daily maximum:", weather_df["max_temperature"].mean())
print("Missing precipitation days:", weather_df["precipitation"].isna().sum())
print("Sum over available days:", weather_df["precipitation"].sum(min_count=1))
```

### Practice and interpretation
Request the same dates for Marquette using latitude `46.54` and longitude `-87.40`. Compare daily maximum temperatures.

Before merging, check that dates, variables, units, and timezones match. A join should have one row per date in each city; duplicated dates could multiply records.

**Discuss:** Can this single month establish a long-term climate trend? Why might a precipitation total be misleading if several days are missing?

**Expected reasoning:** One month is too short for a climate trend. A sum over available days can understate the full-month total. Report coverage and missingness along with the summary.

## 6. World Bank API: country–year observations
[Documentation](https://datahelpdesk.worldbank.org/knowledgebase/articles/898581-api-basic-call-structures)

**Question:** What total population values are reported for Bangladesh and the United States from 2020 to 2023?

The indicator `SP.POP.TOTL` means total population. The country codes are `BGD` and `USA`. The semicolon requests both countries. `format=json` selects JSON.

### Step 1 — Request the indicator

```python
wb_url = "https://api.worldbank.org/v2/country/BGD;USA/indicator/SP.POP.TOTL"
wb_params = {"format": "json", "date": "2020:2023", "per_page": 100}
r = requests.get(wb_url, params=wb_params, timeout=30)
r.raise_for_status()
wb_data = r.json()
print("Top-level type:", type(wb_data).__name__)
print("Metadata:", wb_data[0])
worldbank_retrieved_at = datetime.now(timezone.utc).isoformat()
```

### Step 2 — Separate metadata from observations
This API returns a two-element list: `wb_data[0]` describes pagination; `wb_data[1]` contains the records. Different APIs use different envelopes.

Check the reported page count. If the query expands beyond one page, collect the remaining pages before calling the dataset complete.

```python
metadata, observations = wb_data
if int(metadata["pages"]) != 1:
    raise ValueError("Query spans multiple pages; increase page size or collect remaining pages")
if not isinstance(observations, list):
    raise ValueError("Expected a list of indicator observations")
print("Observations:", len(observations))
if observations:
    print(json.dumps(observations[0], indent=2))
```

### Step 3 — Create a country–year table
Extract nested country names, the year, and the reported value. A null value is missing, not zero. Keep country codes for joining.

```python
population_df = pd.DataFrame([{
    "country_code": obs["countryiso3code"], "country": obs["country"]["value"],
    "year": int(obs["date"]), "population": obs["value"]
} for obs in observations])
population_df["population"] = pd.to_numeric(population_df["population"], errors="coerce")
if population_df.duplicated(["country_code", "year"]).any():
    raise ValueError("Duplicate country-year observations")
display(population_df.sort_values(["country_code", "year"]))
```

```python
population_wide = population_df.pivot(index="year", columns="country_code", values="population").sort_index()
display(population_wide)
(population_wide / 1_000_000).plot(marker="o")
plt.ylabel("Population (millions)")
plt.xlabel("Year")
plt.title("World Bank total population indicator")
plt.xticks(population_wide.index)
plt.tight_layout()
plt.show()
```

### Practice and interpretation
Change the indicator to `NY.GDP.PCAP.CD` (GDP per capita, current US dollars). Update the column name and plot label to match the new variable.

**Discuss:** Why would a population axis label be wrong after changing only the URL? Why might the same historical query produce different values next semester?

**Expected reasoning:** The variable's meaning and units changed. Providers can revise historical estimates; saving the response records the exact data analyzed.

## 7. Reliable collection and saving results
### Errors, authentication, and rate limits
| Status/problem | Action |
|---|---|
| `400` | Check parameter names, types, and allowed ranges |
| `401` / `403` | Check credentials, access, and provider-specific quota information |
| `404` | Check the endpoint path or identifier |
| `429` | Respect provider retry timing; reduce requests |
| `500` / `503` | Consider a bounded retry for a temporary failure |
| JSON decoding failure | Inspect content type; the body may be HTML or empty |
| Missing expected field | Check the schema before creating a table |

API keys identify a client; authorization determines which operations are allowed. Keep personal keys out of notebooks, exported metadata, and printed URLs. NASA's public demonstration key is not a personal credential, but its limited quota can still interrupt a lesson.

Avoid automatically retrying invalid requests. For temporary GET failures, use a small retry limit and backoff. Retrying POST operations can duplicate work unless the provider documents an idempotency mechanism.

### Validate before interpreting
Check required fields, numeric types, units, unique row keys, pagination, and expected date/entity coverage. An absent observation and a returned observation with a null value are different. Do not silently replace either with zero.

### Save each case's DataFrame and collection details
Saving exact inputs supports repeatable analysis even if historical estimates are revised. Keep the raw response when possible, plus the endpoint, nonsecret parameters, collection timestamp, and processing code.

NASA responses can contain links with an API key. Remove `links` fields before exporting its JSON so personal keys cannot leak. The resulting JSON is a sanitized snapshot; weather and World Bank snapshots retain their original parsed structure.

```python
def remove_links(value):
    if isinstance(value, dict):
        return {key: remove_links(item) for key, item in value.items() if key != "links"}
    if isinstance(value, list):
        return [remove_links(item) for item in value]
    return value

snapshots = {"nasa_asteroids": remove_links(nasa), "weather": weather, "worldbank": wb_data}
tables = {"nasa_asteroids": asteroid_df, "weather": weather_df, "worldbank": population_df}
for name in snapshots:
    (output_dir / f"{name}.json").write_text(json.dumps(snapshots[name], indent=2), encoding="utf-8")
    tables[name].to_csv(output_dir / f"{name}.csv", index=False)
collection_notes = {
    "nasa": {"endpoint": nasa_url, "parameters": nasa_dates, "retrieved_at_utc": nasa_retrieved_at},
    "weather": {"endpoint": weather_url, "parameters": weather_params, "retrieved_at_utc": weather_retrieved_at},
    "worldbank": {"endpoint": wb_url, "parameters": wb_params, "retrieved_at_utc": worldbank_retrieved_at}
}
(output_dir / "collection_notes.json").write_text(json.dumps(collection_notes, indent=2), encoding="utf-8")
print("Saved three DataFrames, response snapshots, and collection notes.")
```
