```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from scipy.stats import norm

rng = np.random.default_rng(2026)
plt.style.use('seaborn-v0_8-whitegrid')
```

## 1. From populations to samples

- A **population** is the complete group or process we want to understand.
- A **parameter** is a numerical feature of a population, such as its mean $\mu$.
- A **sample** is the observed subset of the population.
- A **statistic** is a number computed from a sample, such as the sample mean $\bar{x}$.

Because different random samples produce different statistics, inference must account for **sampling variability**.

```python
# A known population: outcomes of one fair six-sided die
population = np.arange(1, 7)
population_mean = population.mean()

small_sample = rng.choice(population, size=3, replace=True)
larger_sample = rng.choice(population, size=25, replace=True)

print('Population mean:', population_mean)
print('n=3 sample:', small_sample, 'mean =', small_sample.mean().round(2))
print('n=25 sample mean:', larger_sample.mean().round(2))
```

### Sampling distribution of the mean

A **sampling distribution** is the distribution of a statistic across many hypothetical random samples of the same size. It is not the distribution of the raw observations.

**Think–pair–share:** What should happen to the sampling distribution of $\bar{x}$ as $n$ increases?

```python
def simulated_sample_means(sample_size, repetitions=10_000):
    samples = rng.choice(population, size=(repetitions, sample_size), replace=True)
    return samples.mean(axis=1)

fig, axes = plt.subplots(1, 3, figsize=(13, 3.5), sharey=True)
for ax, n in zip(axes, [3, 25, 100]):
    means = simulated_sample_means(n)
    ax.hist(means, bins=30, density=True, color='#4C78A8', edgecolor='white')
    ax.axvline(population_mean, color='#E45756', linestyle='--', label=r'$\mu=3.5$')
    ax.set(title=f'n = {n}', xlabel='Sample mean')
axes[0].set_ylabel('Density')
axes[0].legend()
plt.tight_layout()
```

### Central Limit Theorem

For a sufficiently large random sample, the sampling distribution of the mean is approximately normal—even when the population itself is not normal.

$$E(\bar X)=\mu, \qquad SE(\bar X)=\frac{\sigma}{\sqrt{n}}$$

Larger samples give a smaller standard error: estimates vary less from sample to sample.

```python
summary = []
for n in [3, 25, 100]:
    means = simulated_sample_means(n)
    summary.append({
        'n': n,
        'Mean of sample means': means.mean(),
        'Empirical SE': means.std(ddof=1),
        'Theoretical SE': population.std(ddof=0) / np.sqrt(n)
    })
pd.DataFrame(summary).round(4)
```

## 2. Point estimates and confidence intervals

A point estimate gives one best guess. A confidence interval adds a margin of error.

When $\sigma$ is known, a 95% confidence interval for a population mean is

$$\bar{x} \pm 1.96\frac{\sigma}{\sqrt{n}}.$$

In practice, when $\sigma$ is unknown—especially for small samples—we use the sample standard deviation and a $t$ interval. Today, the focus is the interpretation.

```python
# One sample from a normal population with mu=100 and sigma=10
n = 30
sample = rng.normal(loc=100, scale=10, size=n)
xbar = sample.mean()
se = 10 / np.sqrt(n)
ci_95 = (xbar - 1.96 * se, xbar + 1.96 * se)

print(f'Sample mean: {xbar:.2f}')
print(f'95% CI: ({ci_95[0]:.2f}, {ci_95[1]:.2f})')
```

### Correct interpretation

If we repeatedly sampled and built intervals by this method, about 95% of those intervals would contain the true parameter.

It is **not** correct to say that this already-computed interval has a 95% probability of containing the fixed population mean.

```python
# Visualize repeated-interval coverage
mu, sigma, n, repetitions = 100, 10, 30, 100
samples = rng.normal(mu, sigma, size=(repetitions, n))
means = samples.mean(axis=1)
margin = 1.96 * sigma / np.sqrt(n)
lower, upper = means - margin, means + margin
covered = (lower <= mu) & (mu <= upper)

fig, ax = plt.subplots(figsize=(10, 5))
for i in range(repetitions):
    ax.plot([lower[i], upper[i]], [i, i], color='#4C78A8' if covered[i] else '#E45756')
ax.axvline(mu, color='black', linestyle='--', label=r'True $\mu$')
ax.set(xlabel='Interval', ylabel='Repeated sample', title=f'Coverage: {covered.mean():.0%}')
ax.legend()
plt.show()
```

## 3. Hypothesis testing: the simulation recipe

1. State a **null hypothesis** $H_0$ and an **alternative hypothesis** $H_A$.
2. Choose a test statistic whose extreme values favor $H_A$.
3. Compute the observed statistic.
4. Simulate the statistic assuming $H_0$ is true.
5. Compute the **p-value**: the proportion of simulated statistics at least as extreme as the observed statistic.
6. Make a conclusion in context.

A small p-value is evidence **against** $H_0$; it is not the probability that $H_0$ is true.

### Example 1: Is this coin fair?

We flip a coin 100 times and observe 59 heads.

- $H_0$: the coin is fair, $P(\text{heads})=0.5$.
- $H_A$: the coin favors heads, $P(\text{heads})>0.5$.
- Test statistic: number of heads in 100 flips.

Large values support the one-sided alternative.

```python
n_flips = 100
observed_heads = 59
repetitions = 100_000

# Under H0, simulate the number of heads in 100 fair-coin flips.
null_heads = rng.binomial(n=n_flips, p=0.5, size=repetitions)
p_value = np.mean(null_heads >= observed_heads)
p_value
```

```python
fig, ax = plt.subplots(figsize=(9, 4))
bins = np.arange(null_heads.min() - 0.5, null_heads.max() + 1.5)
ax.hist(null_heads, bins=bins, density=True, color='#B8C5D6', edgecolor='white')
ax.hist(null_heads[null_heads >= observed_heads], bins=bins, density=True,
        color='#F58518', edgecolor='white', label=f'p-value ≈ {p_value:.3f}')
ax.axvline(observed_heads, color='#E45756', linestyle='--', label='Observed: 59')
ax.set(xlabel='Heads in 100 flips', ylabel='Density', title='Null distribution')
ax.legend()
plt.show()
```

### Decision and language

At a 5% significance level, a p-value near 0.04 leads us to **reject $H_0$** in favor of the one-sided alternative.

We never “accept” $H_0$. If the p-value is not small, we **fail to reject** it because the data do not provide sufficient evidence against it.

**Check:** How would the alternative and test statistic change if the question were simply “Is the coin biased?”

```python
# Two-sided alternative: the coin may favor either heads or tails.
observed_distance = abs(observed_heads - n_flips / 2)
null_distances = abs(null_heads - n_flips / 2)
two_sided_p = np.mean(null_distances >= observed_distance)
two_sided_p
```

## 4. Comparing categorical distributions with TVD

For categorical distributions $P=(p_1,\ldots,p_k)$ and $Q=(q_1,\ldots,q_k)$, the **total variation distance** is

$$\operatorname{TVD}(P,Q)=\frac{1}{2}\sum_{i=1}^{k}|p_i-q_i|.$$

- TVD = 0 means identical distributions.
- Larger TVD means greater separation.
- The maximum possible TVD is 1.

```python
def tvd(p, q):
    return np.abs(np.asarray(p) - np.asarray(q)).sum() / 2

eth = pd.DataFrame({
    'California': [0.15, 0.05, 0.39, 0.35, 0.06],
    'UCSD':       [0.51, 0.02, 0.16, 0.20, 0.11]
}, index=['Asian', 'Black', 'Latino', 'White', 'Other'])
eth
```

```python
ax = eth.plot.barh(figsize=(8, 4), color=['#4C78A8', '#F58518'])
ax.set(xlabel='Proportion', title='Ethnic distribution: California and UCSD')
plt.show()

observed_tvd = tvd(eth['California'], eth['UCSD'])
print(f'Observed TVD: {observed_tvd:.3f}')
```

### California–UCSD hypothesis test

- $H_0$: UCSD students look like a random sample from California’s ethnic distribution.
- $H_A$: UCSD’s distribution differs from California’s distribution.
- Test statistic: TVD between the observed UCSD and California proportions.

To generate the null distribution, repeatedly sample 30,000 students from the California proportions.

```python
n_students = 30_000
repetitions = 20_000
california = eth['California'].to_numpy()

simulated_props = rng.multinomial(n_students, california, size=repetitions) / n_students
null_tvds = np.abs(simulated_props - california).sum(axis=1) / 2
tvd_p_value = np.mean(null_tvds >= observed_tvd)
tvd_p_value
```

```python
fig, ax = plt.subplots(figsize=(9, 4))
ax.hist(null_tvds, bins=35, color='#4C78A8', edgecolor='white')
ax.axvline(observed_tvd, color='#E45756', linestyle='--', linewidth=2,
           label=f'Observed TVD = {observed_tvd:.3f}')
ax.set(xlabel='TVD', ylabel='Frequency', title='Null distribution of TVD')
ax.legend()
plt.show()

print('Estimated p-value:', tvd_p_value)
```

### Conclusion

None of the simulated TVDs are remotely as large as the observed TVD. The simulation reports 0, meaning only that the p-value is smaller than the simulation’s resolution—not that the true p-value is exactly zero.

We reject $H_0$: the observed UCSD distribution is not plausibly explained by random sampling from the statewide distribution.

**Important:** Statistical significance depends on both effect size and sample size. TVD describes effect size; the test assesses whether chance is a plausible explanation.

## 5. Comparing two groups with a permutation test

The previous test started with a known population distribution. Now we have two observed groups and ask whether their numerical outcomes could come from the same distribution.

### Example 2: Maternal smoking and baby birth weight

Each row is a mother–baby pair. If `data/babyweights.csv` from the source notebook is available, the cell below uses it. Otherwise, it creates a fixed simulated teaching copy with similar group sizes and means so this single notebook runs offline. The fallback is clearly marked and should not be presented as observed study data.

```python
local_baby = Path('data') / 'babyweights.csv'
if local_baby.exists():
    baby = pd.read_csv(local_baby)
    baby_data_source = 'Observed data: data/babyweights.csv'
else:
    offline_rng = np.random.default_rng(2201)
    baby = pd.DataFrame({
        'Maternal Smoker': np.r_[np.ones(459, dtype=bool), np.zeros(715, dtype=bool)],
        'Birth Weight': np.r_[offline_rng.normal(113.8, 18, 459),
                              offline_rng.normal(123.1, 18, 715)]
    })
    baby_data_source = 'SIMULATED offline teaching data'

# Standardize common column spellings used by versions of the dataset.
baby = baby.rename(columns={
    'Birth Weight': 'Birth Weight', 'Birth.Weight': 'Birth Weight',
    'Maternal Smoker': 'Maternal Smoker', 'Maternal.Smoker': 'Maternal Smoker'
})[['Maternal Smoker', 'Birth Weight']].dropna()
baby['Maternal Smoker'] = baby['Maternal Smoker'].astype(bool)
print(baby_data_source)
baby.head()
```

```python
baby.groupby('Maternal Smoker')['Birth Weight'].agg(['count', 'mean', 'std']).round(2)
```

```python
fig, ax = plt.subplots(figsize=(9, 4))
for smoker, color, label in [(False, '#4C78A8', 'Non-smoker'), (True, '#F58518', 'Smoker')]:
    values = baby.loc[baby['Maternal Smoker'] == smoker, 'Birth Weight']
    ax.hist(values, bins=30, density=True, alpha=0.55, color=color, label=label)
ax.set(xlabel='Birth weight (ounces)', ylabel='Density', title='Birth-weight distributions')
ax.legend()
plt.show()
```

### Hypotheses and statistic

- $H_0$: Birth weights have the same distribution for babies of smokers and non-smokers. The group labels are exchangeable.
- $H_A$: Babies of smokers have lower birth weights on average.
- Test statistic:

$$\text{mean weight}_{\text{non-smoker}}-\text{mean weight}_{\text{smoker}}.$$

Large positive values support $H_A$.

```python
def difference_in_means(frame, label_col='Maternal Smoker', value_col='Birth Weight'):
    means = frame.groupby(label_col)[value_col].mean()
    return means.loc[False] - means.loc[True]

observed_difference = difference_in_means(baby)
print(f'Observed difference: {observed_difference:.2f} ounces')
```

### Why shuffle?

Under $H_0$, the smoker/non-smoker label has no relationship with birth weight. Therefore, any reassignment of the existing labels to the existing weights is equally plausible.

Shuffling breaks any real association while preserving:

- every observed birth weight;
- the number of observations in each group; and
- all other features of the pooled outcome distribution.

```python
# One permutation under H0
one_shuffle = baby.assign(
    **{'Maternal Smoker': rng.permutation(baby['Maternal Smoker'].to_numpy())}
)
print('Observed difference:', round(observed_difference, 2))
print('One shuffled difference:', round(difference_in_means(one_shuffle), 2))
```

```python
repetitions = 5_000
permuted_differences = np.empty(repetitions)

for i in range(repetitions):
    shuffled = baby.assign(
        **{'Maternal Smoker': rng.permutation(baby['Maternal Smoker'].to_numpy())}
    )
    permuted_differences[i] = difference_in_means(shuffled)

permutation_p_value = np.mean(permuted_differences >= observed_difference)
permutation_p_value
```

```python
fig, ax = plt.subplots(figsize=(9, 4))
ax.hist(permuted_differences, bins=40, color='#4C78A8', edgecolor='white')
ax.axvline(observed_difference, color='#E45756', linestyle='--', linewidth=2,
           label=f'Observed = {observed_difference:.2f} oz')
ax.set(xlabel='Non-smoker mean − smoker mean (ounces)', ylabel='Frequency',
       title='Permutation null distribution')
ax.legend()
plt.show()

print('Estimated one-sided p-value:', permutation_p_value)
```

### Conclusion—and a crucial limitation

The observed difference is far into the right tail of the permutation null distribution, providing strong evidence of an association between maternal smoking status and birth weight.

However, this is an **observational study**. A small p-value does not by itself establish that smoking caused the difference; confounding variables may also differ between the groups.

## 6. Choosing the simulation strategy

| Question | Null model | Statistic | Simulation |
|---|---|---|---|
| Is a coin fair? | Known categorical probabilities | Heads or distance from 50 | Draw from the null probabilities |
| Does one categorical sample match a known population? | Known population proportions | TVD | Multinomial samples from population |
| Do two observed groups differ? | Labels are exchangeable | Difference in means | Permute group labels |

The common logic is always: **If $H_0$ were true, how often would a statistic at least this extreme occur?**

- Statistical significance is not the same as practical importance.
- Association is not automatically causation.
